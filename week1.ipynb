```python
plt.figure(figsize=(25,10))
plt.title("Categories Having Highest Average Ratings",fontsize=18,fontweight="bold",fontfamily="Tahoma")
sns.barplot(x="Category",y="Rating",data=df,hue="Gender_Target",palette="twilight",legend=True)
plt.show()
```

```python
plt.figure(figsize=(54,20))
plt.title("Brands Having Highest Average ratings",fontsize=18,fontweight="bold",fontfamily="sans-serif")
sns.barplot(x="Brand",y="Rating",data=df,hue="Gender_Target",palette="viridis_r",legend=True)
plt.show()
```

```python
plt.figure(figsize=(16,10))
plt.title("Product Price VS Product Ratings",fontsize=18,fontweight="bold")
sns.scatterplot(x="Price_USD",y="Rating",color="Royalblue",alpha=0.85,data=df)
plt.tight_layout()
plt.show()
pearson_corr=df["Price_USD"].corr(df["Rating"],method="pearson")
spearman_corr=df["Price_USD"].corr(df["Rating"],method="spearman")
def interpret_correlation(corr):
    value = abs(corr)
    if value >= 0.7:
        strength = "Strong correlation"
    elif value >= 0.4:
        strength = "Moderate correlation"
    elif value >= 0.2:
        strength = "Weak correlation"
    else:
        strength = "Very weak or no correlation"
    return f"{strength} correlation"
print("Pearson correlation:", round(pearson_corr, 3))
print("Pearson interpretation:", interpret_correlation(pearson_corr))
print("Spearman correlation:", round(spearman_corr, 3))
print("Spearman interpretation:", interpret_correlation(spearman_corr))
```

```python
plt.figure(figsize=(20,12))
plt.title("Number of Reviews VS Product Ratings",fontsize=18,fontweight="bold")
sns.scatterplot(data=df,x="Number_of_Reviews",y="Rating",alpha=0.5)
plt.tight_layout()
plt.show()
pearson_corr = df["Number_of_Reviews"].corr(df["Rating"], method="pearson")
spearman_corr = df["Number_of_Reviews"].corr(df["Rating"], method="spearman")
def interpret_correlation(corr):
    value = abs(corr)
    if value >= 0.7:
        strength = "Strong correlation"
    elif value >= 0.4:
        strength = "Moderate correlation"
    elif value >= 0.2:
        strength = "Weak correlation"
    else:
        strength = "Very weak or no correlation"
    return f"{strength} correlation"
print("Pearson correlation:", round(pearson_corr, 3))
print("Pearson interpretation:", interpret_correlation(pearson_corr))
print("Spearman correlation:", round(spearman_corr, 3))
print("Spearman interpretation:", interpret_correlation(spearman_corr))
```

```python
plt.figure(figsize=(15,8))
plt.title("Number of Reviews Based on Skin Type",fontsize=18,fontweight="bold")
sns.boxplot(x="Skin_Type",y="Rating",data=df,hue="Gender_Target",palette="ch:start=.2,rot=.3")
plt.legend(loc="upper right")
plt.show()
```

```python
plt.figure(figsize=(20,12))
plt.title("Main Ingredient VS Rating",fontsize=18,fontweight="bold")
sns.barplot(x="Rating",y="Main_Ingredient",data=df)
plt.show()
```

```python
df["Cruelty_Status"]=df["Cruelty_Free"].map({True: "Cruelty-Free",False: "Non-Cruelty-Free"})
fig,axes=plt.subplots(1,2,figsize=(12,6))
sns.boxplot(data=df,x="Cruelty_Status",y="Price_USD",ax=axes[0])
axes[0].set_title("Price VS Cruelty_Status")
sns.boxplot(data=df,x="Cruelty_Status",y="Rating",ax=axes[1])
axes[1].set_title("Rating VS Cruelty_Status")
plt.tight_layout()
plt.show()
```
