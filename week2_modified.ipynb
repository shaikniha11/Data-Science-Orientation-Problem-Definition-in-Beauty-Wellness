```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "most_used_beauty_cosmetics_products_extended.csv"
)
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("\nData types:")
print(df.dtypes)
```

```python

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("\nExact duplicate rows:", df.duplicated().sum())
```

```python

print("\nNumerical summary:")
print(df.describe())
```

```python
print("\nSample records:")
display(df.head())
```

```python
print("Price below or equal to zero:",(df["Price_USD"] <= 0).sum())
print("Ratings outside the 1-5 range:",(~df["Rating"].between(1, 5)).sum())
print("Negative review counts:",(df["Number_of_Reviews"] < 0).sum())
```

```python
categorical_cols = [
    "Brand", "Category", "Skin_Type","Gender_Target", "Main_Ingredient","Country_of_Origin"]
for col in categorical_cols:
    print(f"\n{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts().head(10))
print("\nPrice quantiles:")
print(df["Price_USD"].quantile([0, 0.25, 0.5, 0.75, 0.95, 0.99, 1]))
print("\nReview-count quantiles:")
print(df["Number_of_Reviews"].quantile([0, 0.25, 0.5, 0.75, 0.95, 0.99, 1]))
```
